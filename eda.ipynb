## 1. Introduction and Marketing Analysis

### 1.1 Company Context

 Company A is a wireless telecommunications operator holding years of historical customer, usage, and billing data across roughly 100,000 subscribers - organized into two linked tables (Client.csv and Record.csv) - but currently operating without in-house machine learning capabilities. As competition intensifies across wireless services, leadership recognizes the need to improve margins, subscriber retention, growth, and customer satisfaction, yet lacks a clear, data-driven view of which segment to prioritize or which commercial lever to pull.

 Turning raw telecom records into business impact starts with framing the right analytical question rather than pursuing predictive accuracy for its own sake. While the dataset includes a binary churn indicator—recording whether a subscriber left within 31 to 60 days after the observation date—the underlying behavioral and billing history also supports broader commercial angles, including revenue optimization, usage trends, equipment lifecycle management, and segment-level behavior.

 The mandate of this project is therefore to bridge data science and commercial strategy: combine external market context with exploratory data analysis (EDA) to diagnose Company A's core business bottleneck, define a targeted machine learning task, and translate model outputs into a concrete, verifiable business proposal.

### 1.2 Why this market matters: the Brazil mobile Telecom Landscape

* **A Mature, Saturated, and Concentrated Market**: To ground this proposal in concrete market dynamics, i frame Company A whithin the Brazilian mobile telecommunications market. By September 2026, Brazil reached 281.74 million active mobile connections, following an annual growth rate of 3.7% recorded in Q2 2026. With more active lines than inhabitants, the market is heavily saturated and dominated by three major carriers - VIVO (37.9%), CLARO (33.1%) and TIM (22.4%) - while regional operators and MVQNOs have steadily increased pressure, growing their share from 5.4% in Q3 2025 to 7.2% in 2026. In this environment, organic greenfield growth is limited; operators expand primarily by capturing subscribers from competitors or increasing Average Revenue Per User (ARPU) among existing  customers. Because acquiring a new customer is estimated to be 5 to 25 times more expensive than retaining an existing one, protecting the current subscriber base directly drives profitability.

* **Local Dynamics Reshaping Retention: Regulation, 5G and Plan Migration**:

    Three structural forces in Brazil directly accelerate customer switching and revenue risk:
    *  **Low Switching Barriers (Number Portability)**: Administered by ABR Telecom under Anatel regulation, mobile number portability has surpassed 100 million cumulative transfers since its launch, with 3.8 million carrier switches in the first half of 2026 alone, making it effortless for dissatisfied users to leave.

    * **Shift Toward Postpaid Value**: Excluding M2M/IoT chips, Brazil's standard mobile base (220.92 million lines as of August 2026) is shifting decisively toward higher - ARPU postpaid contracts (128.55 million lines), while prepaid connections continue to shrink (92.37 million lines, declining for consecutive months). Losing a postpaid subscriber therefore carries a much higher lifetime value penalty.

    * **5G Expansion and Equipment Lifecycle**: 5G adoption scale rapidly to 66.1 million lines (23.9% of all mobile accesses) in Q2 2026 and reached 69.8 million accesses shortly after. Customers using aging , non-5G handsets face a growing gap in service experience, making equipment upgrade cycles a critical trigger for either retention or defection.

* **Voluntary vs Involutary Churn: Where Machine Learning creates value**: Globally, telecom churn averages 20% - 50% per year (typically 10% - 20% in postpaid versus to 70% in prepaid), and leading Brazilian operators actively compete to hold postpaid monthly churn near 1.0%. Industry benchmarks distinguish two fundamentally different mechanisms behind subscriber loss:

    * **Involuntary Churn (~12% - 20% of total attrition)**: Driven by non-customer-initiated events such as payment failures, billing friction, credit default, or fraud. This is primarily an operational problem addressed through automated dunning, flexible billing, and credit controls rather than behavioral marketing.

    * **Voluntary Churn (80%+ of total attrition)**: Customer-initiated cancellation triggered by competitive pricing, deteriorating service quality, unresolved support issues, or declining engagement.

* **Connecting Market Reality to Company A**: Behavioral warning signs - such as dropping voice/data usage, billing shocks, customer care calls and aging devices - typically precede voluntary churn and revenue compression by months. While Company A cannot eliminate involuntary shocks like sudden relocation, predictive machine learning models can meaningfully target voluntary churn and hight-value segment erosion, enabling proactive, cost-effective retention before the customer initiates portability.

### 3.4 EDA directly to Business Hypothesis

* **Revenue Concentration (Pareto / High-Value)**: How much of the revenue comes from the top of the base (rev_Mean)? Does it make sense to focus on Revenue-at-Risk (high-value customers)?

* **Bill Shock / Surplus (ovrmou_Mean, totmrc_Mean, rev_Mean)**: Are customers paying significantly above their fixed monthly plan going out more or spending more?

```python
# Correlation/difference between churn=0 and churn=1 in key business variables
candidate_cols = [
    "rev_Mean", "totmrc_Mean", "mou_Mean", "ovrmou_Mean", "ovrrev_Mean",
    "change_mou", "change_rev", "months", "eqpdays", "hnd_price",
    "custcare_Mean", "drop_vce_Mean", "blck_vce_Mean", "comp_vce_Mean",
    "totcalls", "totrev", "avgrev", "avgmou", "models", "phones"
]

available_cols = [c for c in candidate_cols if c in df.columns]
```

```python
summary_by_churn = df.groupby("churn")[available_cols].agg(["median", "mean"]).T
display(summary_by_churn)
```

```python
# Revenue concentration (Rev_Mean quartiles)

df_valid_rev = df[df["rev_Mean"] > 0].copy()
df_valid_rev["rev_quartile"] = pd.qcut(
    df_valid_rev["rev_Mean"], q=4, labels=["Q1 (Low)", "Q2 (Mid-Low)", "Q3 (Mid-High)", "Q4 (High)"]
)
rev_analysis = df_valid_rev.groupby("rev_quartile", observed=False).agg(
    customers=("churn", "count"),
    churn_rate=("churn", "mean"),
    avg_rev=("rev_Mean", "mean"),
    total_rev=("rev_Mean", "sum"),
    avg_eqpdays=("eqpdays", "mean"),
    avg_ovrmou=("ovrmou_Mean", "mean")
)
rev_analysis["rev_share_pct"] = (rev_analysis["total_rev"] / rev_analysis["total_rev"].sum() * 100).round(2)
rev_analysis["churn_rate_pct"] = (rev_analysis["churn_rate"] * 100).round(2)
display(rev_analysis[["customers", "rev_share_pct", "avg_rev", "churn_rate_pct", "avg_eqpdays", "avg_ovrmou"]])
```

```python
# Graphs of the 4 hypotheses (2x2)
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Plot A: Device age (eqpdays) vs. Churn
df_eqp = df[df["eqpdays"] >= 0].copy()
df_eqp["eqp_bin"] = pd.cut(df_eqp["eqpdays"], bins=[0, 180, 365, 540, 730, 2000],
                           labels=["0-6m", "6-12m", "12-18m", "18-24m", "24m+"])
eqp_churn = df_eqp.groupby("eqp_bin", observed=False)["churn"].mean() * 100
sns.barplot(x=eqp_churn.index, y=eqp_churn.values, ax=axes[0, 0], palette="viridis")
axes[0, 0].set_title("Churn Rate by Equipment Age (eqpdays)")
axes[0, 0].set_ylabel("Churn Rate (%)")
axes[0, 0].axhline(df["churn"].mean()*100, color="red", linestyle="--", label="Overall avg")
axes[0, 0].legend()

# Plot B: Time served (months) vs Churn
df["months_bin"] = pd.cut(df["months"], bins=[0, 6, 12, 24, 36, 100],
                          labels=["0-6m", "6-12m", "12-24m", "24-36m", "36m+"])
m_churn = df.groupby("months_bin", observed=False)["churn"].mean() * 100
sns.barplot(x=m_churn.index, y=m_churn.values, ax=axes[0, 1], palette="magma")
axes[0, 1].set_title("Churn Rate by Customer Tenure (months)")
axes[0, 1].set_ylabel("Churn Rate (%)")
axes[0, 1].axhline(df["churn"].mean()*100, color="red", linestyle="--")

# Plot C: Account Spending Limit (asl_flag) & New vs. Refurbished Device (refurb_new)
cat_check = pd.DataFrame({
    "asl_flag=N": [df[df["asl_flag"]=="N"]["churn"].mean()*100],
    "asl_flag=Y": [df[df["asl_flag"]=="Y"]["churn"].mean()*100],
    "New Handset": [df[df["refurb_new"]=="N"]["churn"].mean()*100],
    "Refurb Handset": [df[df["refurb_new"]=="R"]["churn"].mean()*100],
}).T
cat_check.columns = ["churn_rate"]
sns.barplot(x=cat_check.index, y=cat_check["churn_rate"], ax=axes[1, 0], palette="coolwarm")
axes[1, 0].set_title("Churn Rate: Credit Limit (ASL) & Handset Condition")
axes[1, 0].set_ylabel("Churn Rate (%)")
axes[1, 0].tick_params(axis="x", rotation=15)
axes[1, 0].axhline(df["churn"].mean()*100, color="red", linestyle="--")

# Plot D: Usage decline (change_mou) vs Churn (using 5%-95% clipping for clean visualization)
if "change_mou" in df.columns:
    q05, q95 = df["change_mou"].quantile([0.05, 0.95])
    sns.boxplot(data=df[(df["change_mou"] >= q05) & (df["change_mou"] <= q95)],
                x="churn", y="change_mou", ax=axes[1, 1], palette="Set2")
    axes[1, 1].set_title("Change in Minutes of Use (change_mou) by Churn")
    axes[1, 1].axhline(0, color="black", linestyle=":")

plt.tight_layout()
plt.show()
```
